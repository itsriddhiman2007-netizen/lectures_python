q1 . [python basics &data structures] print the resukts of int(4.7), float("3"),bool(0),bool("False") and str(25) +"5" along with the type of each result for bool ("false")

```python
print(int(4.7))
print(type(int(4.7)))

print(float("3"))
print(type(float("3")))

print(bool(0))
print(type(bool(0)))

print(bool("False"))  # True because "False" is a non-empty string
print(type(bool("False")))

print(str(25) + "5")
print(type(str(25) + "5"))
```

q2 . [strings,I/O &Operators] accept a word and check whether it is a palindrome (ignoring case) using string slicing , without using any loop . print the result as a clear message

```python
word = input("Enter a word: ")

if word.lower() == word[::-1].lower():
    print("The word is a palindrome.")
else:
    print("The word is not a palindrome.")
```

q3 . [contol flow & loops] let rolls= [101,104,105,109,112]. Accept a roll number and search for it with a for loop . stop immediately (use break) when it is found and print its position ; otherwise print "not found"

```python
rolls = [101, 104, 105, 109, 112]

num = int(input("Enter roll number: "))

for i in range(len(rolls)):
    if rolls[i] == num:
        print("Found at position:", i + 1)
        break
else:
    print("not found")
```

q4 . [functions] write a function simple_interest(p,r,t=1) that returns the simpel interest . call it with and without time argument and print both the results to 2 decimals

```python
def simple_interest(p, r, t=1):
    return (p * r * t) / 100

print("Simple Interest:", format(simple_interest(5000, 5), ".2f"))
print("Simple Interest:", format(simple_interest(5000, 5, 2), ".2f"))
```
