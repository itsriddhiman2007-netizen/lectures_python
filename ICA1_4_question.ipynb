1> Write a program to calculate Simple Interest, bur first show the WRONG output produced if you accidentally use // instead of / for the final division by 100, then show the CORRECT output

```python

P = 50000
R = 5
T = 2
incorrect_si = P*R*T//100
print("Wrong Simple Interest:" , incorrect_si)
correct_si =P*R*T/100
print("Correct Simple Interest:" ,correct_si)
```

3> Write a program that finds the SUM of only the elmenets at EVEN index positions (0,2,4,....) of a list using slicing

```python
numbers = [10,20,30,40,50]
even_index_sum = sum(numbers[::2])
print("Sum of only even positions :" , even_index_sum)
```

4> Wrrite a program to count the number of DISTINCT (unique) letters in the word "mississippi" using a set .

```python
word = "mississippi"
unique_letters = set(word)
count = len(unique_letters)
print("The number of distinct letters in MISSISSIPPI is :" , {count})
```
