```python
Phones = {
    "iphone 1": {
        "brand": "Apple",
        "operating system": "iOS",
        "cpu speed": "3.2GHZ",
        "Cost": [40000],
        "Memory": [4 + 128]
    },
    "iphone 2": {
        "brand": "Apple",
        "operating system": "iOS",
        "cpu speed": "3.5GHZ",
        "Cost": [55000],
        "Memory": [6 + 256]
    }
}

preference= input("Which iPhone do you prefer? (Type '1' for iphone 1, '2' for iphone 2, or the full name): ").strip().lower()

if preference == '1':
    preference = 'iphone 1'
elif preference == '2':
    preference = 'iphone 2'
else:
    preference = preference

if preference in Phones:
    selected_phone = Phones[preference]
    print(f"Configuration for {preference}:")
    for key, value in selected_phone.items():
        print(f"  {key}: {value}")
else:
    print("Invalid preference")
```
